## Main program
1. Start from an input file (tbtracksXXXX.root). A couple of TTrees only needed.
   * Probably a simulation program? 
2. Define a reader of this TTree:
   * classes to deal with the branches, and incorporates extra functionality (ntuple_utils.py, but simplified)
3. Define a plotter class 
4. Define a fitter class
5. A steering file which is doing the analysis:
   * Read a ntuple 
   * Select a DUT plane
   * Estimate efficiency
   * Obtain resolutions (fitting a gauss or binary uniform)
