```python
# Logic error example
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    average = total / total  # Error lògic: hauria de ser total / count
    # Error: diivdir per la suma en lloc de pel nombre d'elements
    return average
numbers = [10, 20, 30, 40, 50]
avg = calculate_average(numbers)
print(f"La mitjana és: {avg}")
```

```python
# Design error example
def get_weighted_average(scores):
    # Error: no considerar el pes de cada nota
    total_score = sum(scores)
    weighted_average = total_score / len(scores)
    return weighted_average

scores = [80, 90, 70]
results = get_weighted_average(scores)

# error coneptual!!!!
# El càlcul de la mitjana ponderada hauria de considerar pesos específics per a cada nota.
```

### Strings

```python
# Underscores
# Volem crear una funció que introdueixi underscores entre cada caràcter

def add_underscores(word):
    new_word = "_"
    for char in word:
        new_word = char + "_"
    return new_word

phrase = "Hello"
print(add_underscores(phrase))

# AQUÍ HI HA UN ERROR I S'HA D'ARREGLAR
```
