## Debugger de ejemplo: `pdb`

`pdb` es el debugger más usado en Python. No necesita de ninguna instalación y funciona en cualquier entorno.


- Es muy ligero y fácil de usar en cualquier programa.
- Permite inspeccionar variables en tiempo real.
- Permite avanzar paso a paso para analizar el flujo de trabajo del código.
- Ideal para un "debugging" rápido en scripts pequeños o medianos.

---


Se puede activar insertando la siguiente línea en un código:


```python
import pdb; pdb.set_trace()
```

Cuando el programa llega a esa línea se detiene y aparece una interfaz interactiva que permite ejecutar comandos como los siguientes:


- `n` – para ejecutar la siguiente línea (next)
- `s` – para entrar dentro de una función (step)
- `c` – para continuar la ejecución (continue)
- `l` – para listar código alrededor de la línea actual (list)
- `p variable` – para imprimir el valor de una variable (print)
- `q` – para salir del debugger (quit)


````python
# Ejemplo de uso de pdb

def sumar(a, b):
    return a + b

def main():
    x = 2
    y = 3

    # Activamos el debugger
    import pdb; pdb.set_trace()

    # Podemos introducir los comandos:
    #  n (next)
    #  p x  (ver valor de x)
    #  p y
    #  s (step) para entrar en funciones
    z = sumar(x, y)

    # Cuando continúes la ejecución con 'c', verás el resultado final.
    print("Resultado:", z)

main()
````
