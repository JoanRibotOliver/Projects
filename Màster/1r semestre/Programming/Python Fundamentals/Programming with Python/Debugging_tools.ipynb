```python
# Underscores
# Volem crear una funció que introdueixi underscores entre cada caràcter

def add_underscores(word):
    new_word = "_"
    for char in word:
        new_word = char + "_"
    return new_word

phrase = "Hello"
print(add_underscores(phrase))

# AQUÍ HI HA UN ERROR I S'HA D'ARREGLAR
```
