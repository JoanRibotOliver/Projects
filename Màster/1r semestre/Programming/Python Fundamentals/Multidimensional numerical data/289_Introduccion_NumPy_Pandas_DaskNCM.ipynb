## 1.1 ¿Qué es Numpy?

`NumPy` es una libretía fundamental para el cálculo numérico en Python. Permite trabajar con arrays multidimensionales de manera eficiente y ofrece una gran variedad de funciones matemáticas, estadísticas y de álgebra lineal.

## 1.8 Memoria y rendimiento

La forma en que NumPy accede a los datos en memoria, afecta el rendimiento. Hay ciertas herramientas que nos permiten manejar grandes cantidades de datos sin saturar la RAM.

* Strides → cómo se recorren los datos en memoria
    * Resultado típico: (256, 4) → para moverte a la siguiente fila, saltas 256 bytes (64 columnas × 4 bytes/float), y para moverte a la siguiente columna, saltas 4 bytes.
    * Si cambias el orden a Fortran (np.asfortranarray), los strides cambian, favoreciendo acceso por columnas en vez de filas, lo que puede mejorar rendimiento según cómo accedes a los datos.
* Views → operaciones sin copiar datos
    * Una view es una nueva “ventana” a los mismos datos en memoria, sin copiar nada.
    * Ventaja: eficiente en memoria y rápido.
* Order → cómo se almacenan los datos, afectando velocidad
    * permite almacenar matrices en orden C (filas contiguas) o orden Fortran (columnas contiguas).
* Memmap → manejar datasets muy grandes que no caben en RAM ==> **Dask**
    *  Solo se cargan en RAM las partes que necesitas.


````python
# CCD-like stack: (n_files, rows, cols)
N, R, C = 4, 64, 64  # keep small for demonstration
ccd_stack = np.zeros((N, R, C), dtype=np.float32)
print("CCD stack shape:", ccd_stack.shape)
# Create synthetic data
rng = np.random.default_rng(123)
ccd_stack = rng.normal(loc=100.0, scale=5.0, size=(N, R, C)).astype(np.float32)  # raw counts
bias = np.full((R, C), 98.5, dtype=np.float32)   # bias frame
# Subtract bias from all files (broadcasts over leading dimension)
calibrated = ccd_stack - bias  # shape: (N, R, C)



# Inspect strides and memory order
A = np.arange(R*C, dtype=np.float32).reshape(R, C)
print("Strides (bytes):", A.strides)

Af = np.asfortranarray(A)
print("Strides (Fortran order):", Af.strides)

# Save and load
np.save('calibrated_stack.npy', calibrated)
loaded = np.load('calibrated_stack.npy')
print("Loaded shape:", loaded.shape)

# Memory-mapped array (useful for large CCD stacks that don't fit in RAM at once)
mm = np.memmap('big_stack.dat', dtype=np.float32, mode='w+', shape=(8, 128, 128))
mm[:] = rng.normal(100, 3, size=mm.shape)
mm.flush()  # ensure written to disk

# Re-open in read-only mode (no RAM copy)
mm_ro = np.memmap('big_stack.dat', dtype=np.float32, mode='r', shape=(8, 128, 128))
print("Memmap shape:", mm_ro.shape)
````

## 2.1 Crear un DataFrame de eventos
Simulamos una tabla con columnas comunes en análisis de detectores: run, module, ccd, event_id, time, x, y, energy_keV, baseline, temperature, saturated.

````python
rng = np.random.default_rng(2025)

# Configuration
n_runs  = 5
modules = [f'M{m:02d}' for m in range(1, 7)]  # 6 modules
ccds    = ['EXT1', 'EXT2', 'EXT3', 'EXT4']

# keep moderate for classroom performance
n_events = 50_000 

# Build columns
runs = rng.integers(1000, 1000 + n_runs, size=n_events)
mods = rng.choice(modules, size=n_events)
ccd_ids = rng.choice(ccds, size=n_events)

start = np.datetime64('2025-03-01T00:00:00')
# Random seconds over ~2 days
times = start + rng.integers(0, 2*24*3600, size=n_events).astype('timedelta64[s]')

# detector position (y,x) in the CCD
x = rng.integers(0, 2048, size=n_events)
y = rng.integers(0, 4096, size=n_events)
energy      = rng.normal(loc=5.0,   scale=0.7, size=n_events)  # keV
temperature = rng.normal(loc=120.0, scale=0.8, size=n_events)  # K
baseline    = rng.normal(loc=50.0,  scale=1.5, size=n_events)
saturated   = rng.random(n_events) < 0.002  # ~0.2%

# Assemble DataFrame
df = pd.DataFrame({
    'run':    runs,
    'module': mods,
    'ccd':    ccd_ids,
    'event_id': np.arange(n_events, dtype=np.int64),
    'time':   times,
    'x':      x,
    'y':      y,
    'energy_keV':    energy,
    'temperature_K': temperature,
    'baseline':      baseline,
    'saturated':     saturated,
})

# Make some missing values to practice cleaning
mask_nan = rng.random(n_events) < 0.01
df.loc[mask_nan, 'energy_keV'] = np.nan

# ver los ultimos 6
df.tail(6)
````

### category type
En pandas, un tipo category se usa para columnas que tienen pocos valores únicos repetidos muchas veces, como etiquetas o categorías (en nuestro ejemplo, sería `module` o `ccd`).

**Ventajas**:
* **Ahorro de memoria**: pandas almacena internamente solo los valores únicos y usa un código numérico para cada fila
* **Velocidad en operaciones de filtrado, agrupamiento y merge**: las comparaciones y joins se hacen sobre las códigos internos en vez de strings largos
* **Orden y análisis más claro**: se pueden definir categorías ordenadas (`ordered=True`) si los valores tienen un orden natural


````python
%%time
# Memory usage (before optimizing categoricals)
print('Memory (MB) before:', round(df.memory_usage(deep=True).sum() / 1e6, 2))
````

````python
%%time
# Convert repeated labels to category to save memory and speed groupby
for col in ['module', 'ccd']:
    df[col] = df[col].astype('category')
print('Memory (MB) after:', round(df.memory_usage(deep=True).sum() / 1e6, 2))
````

## 2.3 Limpieza: valores faltantes, outliers y flags de calidad
* Eliminar o imputar valores faltantes de energy_keV usando medianas por grupo (por run, module).
* Eliminar valores saturados o físicamente imposibles.
* Mantener un pipeline ordenado y reproducible con encadenamiento de métodos y .pipe().

````python
df[mask_nan].head(3)
````

````python
# Helper to fill NaNs with group medians
def fill_group_median(frame, col, by):
    ### Calcula la mediana de la columna col dentro de cada grupo definido por by
    #  Esto devuelve una Serie del mismo tamaño que el DataFrame, donde cada fila tiene la media de su grupo
    # observed: controla si se incluyen todas las combinaciones posibles de categorías (False) o solo las que realmente existen en los datos (observed=True).
    med = frame.groupby(by, observed=True)[col].transform('median')
    # fillna Rellena los valores faltantes con la mediana correspondiente a su grupo.
    # assign Devuelve un nuevo DataFrame con la columna actualizada. Esto mantiene la cadena de métodos limpia y reproducible.
    return frame.assign(**{col: frame[col].fillna(med)})
````

Veamos 

```bash
        df original
           ↓ rellenar NaNs por mediana
        df_filled
           ↓ eliminar negativos
        df_positive
           ↓ eliminar saturados
        df_valid
           ↓ copiar
        clean
```

````python
clean = (
    df
    .pipe(fill_group_median, 'energy_keV', by=['run', 'module'])
    .query('energy_keV >= 0')
    .query('not saturated')
    .copy()
)

clean[['energy_keV']].describe()
````

## 2.4 Uniones/merges: aplicando calibraciones
* Crearemos tablas de calibración simples y las uniremos a la tabla de eventos:
  * Pedestal a nivel de run: restar de la baseline.
  * Ganancia a nivel de módulo: escalar la energía.


````python
print("columns: ", clean.columns.to_list(), "\n")

# Synthetic calibration tables
runs_unique = np.sort(clean['run'].unique())
modules_unique = clean['module'].cat.categories

pedestal = pd.DataFrame({
    'run': runs_unique,
    'pedestal': 98.5 + 0.05 * (runs_unique - runs_unique.min())  # small drift
})

gain = pd.DataFrame({
    'module': modules_unique,
    'gain': np.linspace(0.98, 1.02, len(modules_unique))
})

# Merge calibrations
#     how{‘left’, ‘right’, ‘outer’, ‘inner’, ‘cross’}, default ‘inner’
cal = (clean
       .merge(pedestal, on='run', how='left')
       .merge(gain,     on='module', how='left')
      )

print("columns: ", cal.columns.to_list(), "\n")

# Add Compute calibrated energy (illustrative)
cal['energy_cal_keV'] = (cal['energy_keV'] - 0.01 * (cal['baseline'] - cal['pedestal'])) * cal['gain']

cal.head()
````

## 2.5 Groupby: estadísticas y eficiencias por run/módulo
* Agregaciones con múltiples estadísticas
* Normalización por grupo usando .transform()
* Eficiencia = fracción que pasa una selección por grupo

````python
# Aggregations
print("Raw columns: ", cal.columns.to_list())
print()
agg = (cal
       .groupby(['run', 'module'])
       .agg(Nevts   =('event_id', 'count'),
            Emean   =('energy_cal_keV', 'mean'),
            Estd    =('energy_cal_keV', 'std'),
            Tmean   =('temperature_K', 'mean'))
      )
print(f"Columns: {agg.columns.to_list()}\n")
agg.head()
````

**Diferencia entre `.agg()`vs `.transform()`**:
* `.agg` → devuelve un *DataFrame por grupo** (resumen)
* `.transform` → devuelve **una columna con el mismo tamaño que el original**, permitiendo operaciones por grupo conservando la estructura original

  

````python
# Per-group z-score using transform
cal['e_z'] = cal.groupby(['run', 'module'])['energy_cal_keV'].transform(lambda s: (s - s.mean())/s.std(ddof=0))
print(f"Columns: {cal.columns.to_list()}\n")
cal[['run','module','energy_cal_keV','e_z']].head()
````

````python
# Efficiency for a physics cut (e.g., signal window)
Emin, Emax = 4.5, 5.5
cal['pass_window'] = cal['energy_cal_keV'].between(Emin, Emax)
cal
````

Se calcula mean() sobre los valores booleanos:
* En Python True = 1, False = 0, por lo que el promedio es la fracción de eventos que pasa el corte → esto es la eficiencia.

````python
eff = cal.groupby(['run', 'module'])['pass_window'].mean().rename('eff_window')
eff.head(3)
````

## 2.7 Series temporales: resampling y estadísticas móviles
* Resamplear la temperatura en intervalos de 15 minutos para monitorear estabilidad.
* Media móvil de energía calibrada por módulo vs tiempo.

````python
# Ensure time is datetime64[ns]
cal['time'] = pd.to_datetime(cal['time'])

# Resample temperature by 15 min per module
temp_ts = (cal
           .set_index('time')
           .groupby('module')['temperature_K']
           .resample('15min')
           .mean()
          )

print(temp_ts.head())
````

````python

# 1) Group efficiency and ranking
Emin, Emax = 4.8, 5.2
win = cal['energy_cal_keV'].between(Emin, Emax)
eff_run_mod = cal.assign(pass_window=win).groupby(['run','module'], observed=True)['pass_window'].mean()
rank = eff_run_mod.groupby('module', observed=True).median().sort_values(ascending=False)
print('Median efficiency by module (descending):')
print(rank)

# 2) Time stability per module (daily resample)
daily = (cal.set_index('time')
           .groupby('module', observed=True)['energy_cal_keV']
           .resample('24h').mean()
           .dropna())
stab = daily.groupby('module', observed=True).std().rename('std_daily_mean')
print('Most stable (lowest std):')
print(stab.sort_values().head())

# 3) Per-CCD correction and recompute
offsets = pd.DataFrame({'ccd': cal['ccd'].cat.categories, 'ccd_offset': [0.00, 0.02, -0.01, 0.01]})
cal2 = cal.merge(offsets, on='ccd', how='left').copy()
cal2['energy_cal_keV2'] = cal2['energy_cal_keV'] + cal2['ccd_offset']
print(cal2[['ccd','ccd_offset','energy_cal_keV','energy_cal_keV2']].head())

# 4) Pivot heatmap of mean e_z (print pivot)
pivot_ez = pd.pivot_table(
    cal, 
    values='e_z', 
    index='module', 
    columns='ccd', 
    aggfunc='mean',
    observed=True
)
print(pivot_ez)

# 5) Method chaining pipeline from raw df
from pandas.api.types import CategoricalDtype

def pipeline(frame):
    modules = CategoricalDtype(categories=sorted(frame['module'].unique()))
    ccds = CategoricalDtype(categories=['EXT1','EXT2','EXT3','EXT4'])
    return (
        frame
        .assign(module=lambda f: f['module'].astype(modules),
                ccd=lambda f: f['ccd'].astype(ccds))
        .pipe(fill_group_median, 'energy_keV', by=['run', 'module'])
        .query('not saturated and energy_keV >= 0')
        .merge(pedestal, on='run', how='left')
        .merge(gain, on='module', how='left')
        .assign(energy_cal_keV=lambda f: (f['energy_keV'] - 0.01 * (f['baseline'] - f['pedestal'])) * f['gain'])
        .assign(e_z=lambda f: f.groupby(['run','module'], observed=True)['energy_cal_keV'].transform(lambda s: (s - s.mean())/s.std(ddof=0)))
        .query('4.5 <= energy_cal_keV <= 5.5')
    )

result = pipeline(df)
print(result.head())
print('Pipeline rows:', len(result))
````

Usa np.linalg.lstsq para ajustar una línea (y = m x + b) a los datos, algo común cuando se ajusta una proyección de trayectoria o una curva de calibración.

````python
import matplotlib.pyplot as plt

# Generate synthetic points along a line with noise
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 50)
y_true = 2.5 * x + 1.0
noise = rng.normal(0, 1.0, size=x.shape)
y = y_true + noise

# Design matrix [x, 1]
A = np.vstack([x, np.ones_like(x)]).T
m, b = np.linalg.lstsq(A, y, rcond=None)[0]
print(f"Fit slope m={m:.3f}, intercept b={b:.3f}")

# Plot (optional)
plt.figure()
plt.scatter(x, y, s=20, label='data')
plt.plot(x, y_true, 'k--', label='true')
plt.plot(x, m*x + b, 'r', label='fit')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Least-squares line fit')
plt.legend()
plt.show()
````

# ¿Que es Dask?

---

Dask es una **biblioteca** de código abierto para Python que permite de manera flexible realizar **computación paralela** o procesamientos de información en paralelo

---

Está compuesta por dos partes:
* Programación dinámica de tareas (**Dynamic task scheduling**) optimizada para el cálculo. Consta de una orquestración de los procesos que queremos realizar, por parte de la libreria, para optimizar el cálculo reduciendo el uso de memoria y reduciendo el tiempo computacional.
* Serie de **Big Data collections** como *parallel arrays*, *dataframes* o *listas* (tal como Numpy, Pandas, etc.).

Estas colecciones se ejecutan sobre estos programadores de tareas dinamicos.

---

Resumiendo:
* Biblioteca de Python para **computación paralela y distribuida**
* **Divide los datos en bloques** (particiones manejables)
* Construye un **grafo de tareas** con operaciones y dependencias
* **Evaluación diferida**: nada se ejecuta hasta llamar a `.compute()`
* **Escalable**: el mismo código funciona en un portátil → multicore → clúster

# 8. Dask Delayed

Muchas veces lo que queremos resolver no "encaja" en colecciones como `array` o `dataframe`. En estos casos, podemos **paralelizar código python usando la API de Dask Delayed**.

Son `lazy`, no se van a ejecutar hasta que sea necesario, o llamemos a `compute`.
    
Veamos algunos ejemplos:

````python
from dask.distributed import Client, LocalCluster

cluster = LocalCluster(
    n_workers=4,            # try 4–8 depending on your machine
    threads_per_worker=1,   # 1 is often best for NumPy-like workloads
    dashboard_address=":8787"
)
client = Client(cluster)
client
````

````python
from time import sleep

def inc(x):
    sleep(.5)
    return x + 1

def double(x):
    sleep(.3)
    return x * 2

def add(x,y):
    sleep(1)
    return x + y
````

````python
%%time
data = np.arange(1,5)

output = []
for x in data:
    a = inc(x)
    b = double(x)
    c = add(a,b)
    output.append(c)
    
total = sum(output)
````

## Probemos esto con Delayed

````python
%%time
from dask import delayed

output = []
for x in data:
    a = delayed(inc)(x)
    b = delayed(double)(x)
    c = delayed(add)(a,b)
    output.append(c)
    
total = delayed(sum)(output)
````

# Orchestration

````python
%%time
total.compute()
````

## Usar Delayed como decorador

Usarlo para aquellas funciones que quieres que sean `lazy`.

````python
import dask 

@dask.delayed
def inc(x):
    sleep(.5)
    return x + 1

@dask.delayed
def double(x):
    sleep(.3)
    return x * 2
    
@dask.delayed
def add(x,y):
    sleep(1)
    return x + y

data = np.arange(1,5)

output = []
for x in data:
    a = inc(x)
    b = double(x)
    c = add(a,b)
    output.append(c)
    
total = sum(output)
````

````python
%%time
total.compute()
````
