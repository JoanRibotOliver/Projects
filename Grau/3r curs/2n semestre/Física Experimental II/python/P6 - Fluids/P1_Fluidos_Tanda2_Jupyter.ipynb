```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

# Define sigma_0 value (sigma0 -> K)
K_N = 0.07275

# Define the logarithmic function
def logarithmic_func(x, alpha, beta):
    return K_N - alpha * np.log10(1 + beta * x)

# Read data from Excel file
file_name = 'excel_P6_1.xlsx'
sheet_index = 1  # Index of the 'Datos_ajuste' sheet (assuming it's the second sheet, index starts from 0)

data = pd.read_excel(file_name, sheet_name=sheet_index)

x_data = data['c'].values
y_data_N = data['sigma'].values
x_error = data['error_c'].values
y_err_N = data['error_sigma'].values
y_error_N = np.array(y_err_N)

# Curve fitting with adjusted initial parameter guesses
params, cov = curve_fit(logarithmic_func, x_data, y_data_N, p0=[0.01, 0.01])

# Best fitting parameters
alpha_fit, beta_fit = params

# Parameter errors (standard deviations)
alpha_err, beta_err = np.sqrt(np.diag(cov))

# Adjusted values
x_fit = np.linspace(min(x_data), max(x_data), 100)
y_fit = logarithmic_func(x_fit, alpha_fit, beta_fit)

# Data and fitted curve plot with error bars
plt.errorbar(x_data, y_data_N, xerr=x_error, yerr=y_error_N, fmt='o', label='Dades', color='C0', linewidth=1, markersize=2, capsize=2)
plt.plot(x_fit, y_fit, label='Ajust logarítmic', linestyle='dashed', color='r', linewidth=2)
plt.xlabel(r'$c$ (v/v)')
plt.ylabel(r'$\sigma$ (N/m)')
plt.xlim(right=1.02)
plt.legend()
plt.title(r'Tensió superficial Vs. Concentració en volum')
plt.grid()
plt.savefig("sigma_vs_c_sin_ajuste_sigma0_python",dpi=300, bbox_inches= "tight")
plt.clf()
plt.show()

alpha_ajuste = alpha_fit
err_alpha_ajuste = alpha_err

beta_ajuste = beta_fit
err_beta_ajuste = beta_err

print("########## Datos con sigma_0 fijo ##########")
print("sigma0:", K_N)
print(f"Best fitting alpha: {alpha_fit:.4f} +/- {alpha_err:.4f}")
print(f"Best fitting beta: {beta_fit:.4f} +/- {beta_err:.4f}")

# Calculate the R^2 correlation factor
residuals = y_data_N - logarithmic_func(x_data, alpha_fit, beta_fit)
ss_residuals = np.sum(residuals**2)
ss_total = np.sum((y_data_N - np.mean(y_data_N))**2)
r_squared = 1 - (ss_residuals / ss_total)

print(f"R^2 factor: {r_squared:.4f}")
print("########################################")
```
